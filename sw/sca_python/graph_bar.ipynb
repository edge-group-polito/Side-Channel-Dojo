```python
import matplotlib.pyplot as plt
import numpy as np

# X-axis labels (S-box implementations)
sboxes = ["standard", "freyre_1", "freyre_2", "freyre_3", "hussain_6", "ozkaynak_1"]

# Data from the left plot (software AES)
software_traces = [575, 1000, 1200, 525, 525, 500]

# Data from the right plot (hardware AES)
hardware_traces = [800, 2325, 1725, 850, 625, 600]

x = np.arange(len(sboxes))       # the label locations
width = 0.35                     # width of each bar

fig, ax = plt.subplots(figsize=(8, 6))

# Bars: blue for software, red for hardware
rects1 = ax.bar(x - width/2, software_traces, width,
                label="SR 90%", color="#E9D8A6")
rects2 = ax.bar(x + width/2, hardware_traces, width,
                label="SR 100%", color="#AE2012")

# Axis labels and ticks
ax.set_ylabel("Number of traces", fontsize=15)
ax.set_xlabel("S-box", fontsize=15)
ax.set_xticks(x)
ax.set_xticklabels(sboxes, rotation=30, ha="right")
ax.tick_params(axis='y', labelcolor='black', labelsize=10)
ax.tick_params(axis='x', labelsize=10)

# Legend
ax.legend()

# Optional: add values on top of each bar
def autolabel(rects):
    """Attach a text label above each bar in *rects*, displaying its height."""
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f"{height}",
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha="center", va="bottom", fontsize=9)

autolabel(rects1)
autolabel(rects2)

fig.tight_layout()
plt.savefig("SR_AES_SW.pdf")
plt.show()
plt.close()
```
